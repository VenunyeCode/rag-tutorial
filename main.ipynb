```python
from pathlib import Path
from pypdf import PdfReader


DATA_DIR = Path("data")

def load_pdfs(directory: Path) -> str:
    documents = []

    for pdf_path in sorted(directory.glob("*.pdf")):
        print(f"Reading: {pdf_path.name}")

        reader = PdfReader(pdf_path)

        for page in reader.pages:
            text = page.extract_text()

            if text:
                documents.append(text)

    return "\n".join(documents)

def chunk_text(text: str, chunk_size: int = 1024) -> list[str]:
    chunks = []

    for start in range(0, len(text), chunk_size):
        chunk = text[start:start + chunk_size]
        if chunk.strip():
            chunks.append(chunk)

    return chunks

#Step 0: Load and parse the documents
text = load_pdfs(DATA_DIR)
chunks = chunk_text(text)

print(f"Extracted characters: {len(text)}")
print(f"Number of chunks: {len(chunks)}")
print(chunks[0])
print(f"Largest chunk: {max(map(len, chunks))}")
```

```python
import numpy as np
from langchain_ollama import OllamaEmbeddings

embedding_model = OllamaEmbeddings(model="mxbai-embed-large")

chunk_embeddings = embedding_model.embed_documents(chunks)

chunk_vectors = np.array(chunk_embeddings)

print("Embeddings shape:", chunk_vectors.shape)
```
